```python
from elasticsearch import Elasticsearch
import pandas as pd

# 连接 ES（本地或集群）
es = Elasticsearch("https://localhost:9200", basic_auth=("elastic", "elastic"), verify_certs=False)

# 查询数据
res = es.search(index="youtube-videos-life", body={"query": {"match_all": {}}}, size=1000)
data = [doc["_source"] for doc in res["hits"]["hits"]]

# 转为 DataFrame
df = pd.DataFrame(data)
df.head()
```
