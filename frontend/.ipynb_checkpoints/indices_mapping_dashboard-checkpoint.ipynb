```python
from elasticsearch import Elasticsearch
import json
import urllib3
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

# Connect to Elasticsearch
es = Elasticsearch(
    hosts=["https://localhost:9200"],
    basic_auth=("elastic", "elastic"),
    verify_certs=False
)

# Define the index name
index_name = "youtube-videos-election" 

# Fetch mapping
mapping = es.indices.get_mapping(index=index_name)



def show_index_mapping(es_client, index_name):
    """
    Print the mapping of a given Elasticsearch index in a readable format.

    Parameters:
    - es_client: an instance of Elasticsearch client
    - index_name: name of the index (string)
    """
    try:
        mapping = es_client.indices.get_mapping(index=index_name)
        print(f"\n=== Mapping for index: {index_name} ===")
        print(json.dumps(mapping[index_name]['mappings'], indent=2))
    except Exception as e:
        print(f"[X] Failed to fetch mapping for index '{index_name}': {e}")
```

```python
show_index_mapping(es, "youtube-videos-election")
```
